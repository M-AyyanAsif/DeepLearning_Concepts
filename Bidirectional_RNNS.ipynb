```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,Dense,SimpleRNN,Bidirectional,LSTM,GRU,Input,Dropout
```

```python
(x_train,y_train),(x_test,y_test)= imdb.load_data(num_words=10000)
```

```python
x_train= pad_sequences(x_train, maxlen=1000)
x_test= pad_sequences (x_test, maxlen=1000)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, Bidirectional, LSTM, Dropout, Dense

model=Sequential([
    Input(shape=(1000,)),
    Embedding(10000,32),
    Bidirectional(LSTM(64, return_sequences=True)),
    Dropout(0.3),
    Bidirectional(LSTM(32)),
    Dense(1,activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss="binary_crossentropy",metrics=["accuracy"])
```

```python
model.fit(x_train,y_train,epochs=5,validation_split=(0.4),batch_size=32)
```
