```python
import numpy as np
words = [
    "Learning data science requires consistency and curiosity.",
    "Python is a powerful language for data analysis and machine learning.",
    "Clean data is more important than complex models.",
    "Practice projects help turn theory into real skills.",
    "Deep learning models perform better with more quality data.",
    "Understanding fundamentals makes advanced topics easier.",
    "Visualization helps communicate insights effectively.",
    "Cloud platforms like AWS are valuable for data engineers.",
    "Hyperparameter tuning can significantly improve model performance.",
    "Continuous learning is key to long-term success in tech."
]
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer(oov_token='<nothing>')
```

```python
tokenizer.fit_on_texts(words)
```

```python
sequences = tokenizer.texts_to_sequences(words)
sequences
```

```python
from keras.utils import pad_sequences
sequences=pad_sequences(sequences, padding='post')
```

```python
sequences
```

```python
from keras.datasets import imdb
from keras.models import Sequential
from keras.layers import Dense,SimpleRNN,Flatten
```

```python
(x_train,y_train),(x_test,y_test)= imdb.load_data()
```

```python
x_train= pad_sequences(x_train,padding="post")
x_test=pad_sequences(x_test,padding="post")
```

```python
x_train[0]
```

```python
model=Sequential()
model.add(SimpleRNN(32,input_shape=(None,1),return_sequences=False))
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```
