```python
import numpy as np
words = [
    "Learning data science requires consistency and curiosity.",
    "Python is a powerful language for data analysis and machine learning.",
    "Clean data is more important than complex models.",
    "Practice projects help turn theory into real skills.",
    "Deep learning models perform better with more quality data.",
    "Understanding fundamentals makes advanced topics easier.",
    "Visualization helps communicate insights effectively.",
    "Cloud platforms like AWS are valuable for data engineers.",
    "Hyperparameter tuning can significantly improve model performance.",
    "Continuous learning is key to long-term success in tech."
]
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer= Tokenizer()
```

```python
tokenizer.fit_on_texts(words)
```

```python
len(tokenizer.word_index)
```

```python
sequences= tokenizer.texts_to_sequences(words)
sequences
```

```python
from tensorflow.keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,Dense,Flatten
```

```python
model= Sequential()
model.add(Embedding(68,output_dim=3,input_length=11))
model.add(Flatten())
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

```python
from keras.datasets import imdb
from keras.layers import Embedding,Flatten,Dense,SimpleRNN
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.models import Sequential
from keras.utils import pad_sequences
```

```python
(x_train,y_train),(x_test,y_test)= imdb.load_data()
```

```python
x_train= pad_sequences(x_train,padding='post')
x_test=pad_sequences(x_test,padding='post')
```

```python
model=Sequential()
model.add(Embedding(10000,32))
model.add(SimpleRNN(32))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(x_train, y_train,epochs=5,validation_data=(x_test,y_test))
```
